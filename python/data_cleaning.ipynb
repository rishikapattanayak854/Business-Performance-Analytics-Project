```python
import pandas as pd

# Load all three datasets
customers = pd.read_csv("../Dataset/customers.csv")
sales = pd.read_csv("../Dataset/sales.csv")
inventory = pd.read_csv("../Dataset/inventory.csv")

print("All datasets loaded successfully!")

print("\nCustomers Dataset:")
display(customers.head())

print("\nSales Dataset:")
display(sales.head())

print("\nInventory Dataset:")
display(inventory.head())
```

```python
product_sales = pd.merge(
    sales,
    inventory[['product_id', 'product_name']],
    on='product_id'
)

top_products = product_sales.groupby(
    'product_name'
)['quantity'].sum().sort_values(
    ascending=False
).head(5)

display(top_products)
```

```python
print("Inventory Columns:")
print(inventory.columns.tolist())

print("\nSales Columns:")
print(sales.columns.tolist())
```

```python
top_products = sales.merge(
    inventory[['product_id', 'product_name']],
    on='product_id',
    how='left'
)

print(top_products.columns.tolist())

top_products = top_products.groupby(
    'product_name'
)['quantity'].sum().reset_index()

top_products = top_products.sort_values(
    by='quantity',
    ascending=False
).head(5)

print(top_products)
```

```python
plt.figure(figsize=(9,5))

plt.bar(
    top_products['product_name'],
    top_products['quantity']
)

plt.title('Top 5 Best-Selling Products')
plt.xlabel('Product Name')
plt.ylabel('Quantity Sold')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
sales['sale_date'] = pd.to_datetime(sales['sale_date'])

monthly_sales = sales.groupby(
    sales['sale_date'].dt.to_period('M')
)['total_amount'].sum().reset_index()

monthly_sales['sale_date'] = monthly_sales['sale_date'].astype(str)

print(monthly_sales)
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    monthly_sales['sale_date'],
    monthly_sales['total_amount'],
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
customer_analysis = sales.groupby('customer_id').agg(
    total_orders=('sale_id', 'count'),
    total_quantity=('quantity', 'sum'),
    total_spent=('total_amount', 'sum')
).reset_index()

customer_analysis = customer_analysis.sort_values(
    by='total_spent',
    ascending=False
)

print(customer_analysis)
```

```python
top_customers = customer_analysis.sort_values(
    by='total_spent',
    ascending=False
).head(5)

plt.figure(figsize=(9,5))

plt.bar(
    top_customers['customer_id'].astype(str),
    top_customers['total_spent']
)

plt.title('Top 5 Customers by Total Spending')
plt.xlabel('Customer ID')
plt.ylabel('Total Amount Spent')

plt.tight_layout()
plt.show()
```

```python
inventory_analysis = inventory.groupby('category').agg(
    total_products=('product_id', 'count'),
    total_stock=('stock_quantity', 'sum')
).reset_index()

print(inventory_analysis)
```

```python
plt.figure(figsize=(9,5))

plt.bar(
    inventory_analysis['category'],
    inventory_analysis['total_stock']
)

plt.title('Category-wise Inventory Stock')
plt.xlabel('Product Category')
plt.ylabel('Total Stock Available')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
